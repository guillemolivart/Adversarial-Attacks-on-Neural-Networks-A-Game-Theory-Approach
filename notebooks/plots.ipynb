```python
import numpy as np
import matplotlib.pyplot as plt
from keras.datasets import mnist

from models import build_mlp_model, build_cnn_model

def plot_predictions(model, X_test, y_test, X_test_original, num_images=10):
    """Funció auxiliar per visualitzar si el model encerta o falla."""
    pred_y_test_prob = model.predict(X_test)
    pred_y_test = np.argmax(pred_y_test_prob, axis=1)
    
    for i in range(num_images):
        pred = pred_y_test[i]
        real = y_test[i]
        
        if pred == real:
            color_title = 'green'
            message = f"Correct! (Prediction: {pred})"
        else:
            color_title = 'red'
            message = f"ERROR! Prediction: {pred} | Real: {real}"
            
        plt.figure(figsize=(2, 2)) 
        plt.imshow(X_test_original[i], cmap='gray') 
        plt.title(message, color=color_title)
        plt.axis('off') 
        plt.show()
```
